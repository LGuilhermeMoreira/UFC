# Predições com múltiplos pontos
- Realizar predições acuradas se torna mais difícil com múltiplos pontos
- Para qualquer conjunto de pesos, existe muitos valores de erro
- ... correspondendo para muitos pontos para os quais as predições foram feitas



# Loss Function 
- Agrega os erros em predições de muitos pontos de dados em apenas um número
- É uma métrica da capacidade de predição do modelo

# Loss Function
- Valor baixo para a Loss Function significa um melhor modelo
- Objetivo: encontrar os pesos que geram o menor valor para a loss function
- Gradient descent ou Método do Gradiente

# Gradient Descent
- Imagine que vc está em um campo escuro
- Deseja encontrar o pnto mais baixo
- Sinta o piso para ver como ele cresce
- Tome um pequeno passo na descendente
- Repita até que seja subida em todas as direções

# Passos do Gradient Descent
- Comece em um ponto randomico
- Repita até estar em algun lugar plano:
    - Encontre uma inclinação
    - Tome um passo na descendente

# Gradient Descent

- Se o pico for positivo:
    - Indo na direção oposta do pico significa mover-se para números menores
    - Subtrair o pico do valor corrente
    - Passo muito grande pode levar a derviar
- Solução: learning rate
    - Atualizar cada peso subtraindo **learning rate * pico**

- Para calcular o pico para um peso, é necessário multiplicar:
    - Pico da loss function em respeito ao nó que é alimentado 
    - Para a loss function sendo o mean squared error seria 2 * error
    - O valor do nó que que alimenta o peso
    - Pico da função de ativação em respeito ao valor que nós alimentamos

# Código para calcular picos e atualizar pesos

```python
import numpy as np
```

```python
weights = np.array([1, 2])
```

```python
input_data = np.array([3, 4])
```

```python
target = 6
```

```python
learning_rate = 0.01
```

```python
preds = (weights * input_data).sum()
```

```python
error = preds - target
```

```python
print(error)
```

```python
gradient = 2 * input_data * error
```

```python
gradient
```

```python
weights_updated = weights - learning_rate * gradient
```

```python
preds_updated = (weights_updated * input_data).sum()
```

```python
error_updated = preds_updated - target
```

```python
print(error_updated)
```

# Backpropagation

- Usa o gradient descent para atualizar todos os pesos de uma rede neural (obtendo gradientes para todos os pesos)
## Processo
- Tenta estimar o pico da Loss function em respeito a cada peso
- Realiza forward propagation para calcular predições e erros
