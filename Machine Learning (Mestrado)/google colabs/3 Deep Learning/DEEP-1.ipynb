# Forward Propagation

- Processo de multiplicação e adição
- A propagação é para uma observação de cada vez
- Saída é a predição para aquela observação específica

```python
import numpy as np
```

```python
input_data = np.array([2,3])
```

```python
weights = {'node_0': np.array([1, 1]), 'node_1': np.array([-1, 1]),
           'output': np.array([2, -1])}
```

```python
node_0_value = (input_data * weights['node_0']).sum()
node_1_value = (input_data * weights['node_1']).sum()
```

```python
hidden_layer_values = np.array([node_0_value, node_1_value])
```

```python
print(hidden_layer_values)
```

```python
output = (hidden_layer_values * weights['output']).sum()
```

```python
print(output)
```

# Função de Ativação
- São aplicadas aos nós de entrada para produzir valores nos nós de saída

```python
import numpy as np
```

```python
input_data = np.array([-1, 2])
```

```python
weights = {'node_0': np.array([3, 3]), 'node_1': np.array([1, 5]),
'output': np.array([2, -1])}
```

```python
node_0_input = (input_data * weights['node_0']).sum()
```

```python
node_0_output = np.tanh(node_0_input)
```

```python
node_1_input = (input_data * weights['node_1']).sum()
```

```python
node_1_output = np.tanh(node_1_input)
```

```python
hidden_layer_outputs = np.array([node_0_output, node_1_output])
```

```python
output = (hidden_layer_outputs * weights['output']).sum()
```

```python
print(output)
```
